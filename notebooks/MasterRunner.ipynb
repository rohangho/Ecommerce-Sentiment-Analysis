## 2. Feature Extraction with Ollama

Using a locally running LLaMA 3 model to extract product features from the product name.

```python
# Load the OllamaFeatureExtractor class
%run FeatureExtraction.ipynb

# Instantiate the extractor with the local LLaMA 3 model
extractor = OllamaFeatureExtractor(model_name="llama3:latest")

# Run feature extraction on a sample of 5 products
# Providing the column 'name' instead of 'reviews.text'
results_df = extractor.analyze_dataframe(eda.df, text_column='name')
display(results_df)
```
